```java
//| echo: true
//| output: true
class Person {
  // Attributs (champs)
  private String name;
  private int age;

  // Constructeur
  public Person(String name, int age) {
    this.name = name;
    this.age = age;
  }

  // Méthodes (comportements)
  public void greet() {
    System.out.println("Bonjour, je suis " + name + ", " + age + " ans.");
  }

  public boolean isAdult() {
    return age >= 18;
  }

  // Getters
  public String getName() { return name; }
  public int getAge() { return age; }
}

// Utilisation
Person alice = new Person("Alice", 29);
Person bob = new Person("Bob", 16);

alice.greet();
bob.greet();
System.out.println(alice.getName() + " est majeur: " + alice.isAdult());
System.out.println(bob.getName() + " est majeur: " + bob.isAdult());
```

```java
//| echo: true
//| output: true
class Animal {
  protected String nom;

  public Animal(String nom) {
    this.nom = nom;
  }

  public void faireDuBruit() {
    System.out.println(nom + " fait du bruit");
  }
}

class Chien extends Animal {
  public Chien(String nom) {
    super(nom);
  }

  @Override
  public void faireDuBruit() {
    System.out.println(nom + " aboie: Ouaf ouaf !");
  }
}

class Chat extends Animal {
  public Chat(String nom) {
    super(nom);
  }

  @Override
  public void faireDuBruit() {
    System.out.println(nom + " miaule: Miaou !");
  }
}

// Polymorphisme
Animal[] animaux = {
  new Chien("Rex"),
  new Chat("Félix"),
  new Chien("Max")
};

for (Animal a : animaux) {
  a.faireDuBruit();
}
```
