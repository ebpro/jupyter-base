```java
String checkAge(int age) {
    if (age >= 18) {
        return "Majeur";
    } else {
        return "Mineur";
    }
}

// Test cases
System.out.println("Age 15: " + checkAge(15));
System.out.println("Age 18: " + checkAge(18));
System.out.println("Age 25: " + checkAge(25));
```

```java
int sumArray(int[] arr) {
    int sum = 0;
    for (int value : arr) {
        sum += value;
    }
    return sum;
}

// Test case
int[] numbers = {10, 20, 30, 40, 50};
System.out.println("Sum: " + sumArray(numbers));
```
